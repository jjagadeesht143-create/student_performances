## Cleaning steps applied

1. **Drop exact duplicate rows** - a duplicate row is a data-entry artifact, not new information, so it is safe to remove before any train/test split.
2. **Drop rows with a missing target (`final_score`)** - we cannot train or evaluate on a row whose label is unknown, and imputing a label would mean fabricating ground truth.
3. **Enforce sane dtypes** for numeric and categorical columns.

Deliberately **NOT** done here (to avoid leakage):
- Imputing missing *feature* values using dataset-wide statistics.
- Scaling or encoding features.

Those steps happen later, inside a scikit-learn `Pipeline` fit only on the training split.

```python
cleaned = clean_data(df)
cleaned.head()
```
